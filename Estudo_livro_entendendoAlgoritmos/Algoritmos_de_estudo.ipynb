# Algoritmo ordenação por seleção

```python
def buscaMenor(arr):
    menor = arr [0]
    menor_indice = 0
    for i in range(1, len(arr)):
        if arr[i] < menor :
            menor = arr[i]
            menor_indice = i 
    return menor_indice
def ordenacaoporSelecao(arr):
    novoArr = []
    for i in range (len(arr)):
        menor = buscaMenor(arr)
        novoArr.append(arr.pop(menor))
    return novoArr

print (ordenacaoporSelecao([5,3,6,2,10,4,7,8,9,1]))
```

# Recursão

```python
# Exemplo de código recursão procurando chaves em uma caixa 
def procure_pela_chave(caixa):
    for item in caixa:
        if item.e_uma_caixa():
            procure_pela_chave(item):
        elif item.e_uma_caixa():
            print("achei a chave")
```

```python
# outro exemplo de recursão 
def fatorial(x):
    if x == 1:
        return 1 
    else :
     return  x * (x-1)
fatorial(5)
```

# Quicksort

função que soma os elementos de uma lista 

```python
def soma(lista):
    total = 0
    for i in lista:
        total += i
    return total 
soma([10,10,10,10,10])
```

função recursiva que conta o numero de itens em uma lista 

```python
def contando(lista):
        if lista == []:
         return 0
        return 1 + contando(lista[1:])
print(contando([1,2,3,4,5]))
```

```python
# recursão soma de elementos de uma lista 
def soma(lista):
    if lista == []:
        return 0 
    else :
        return lista[0] + soma(lista[1:])
soma([1,2,3,4])
    
```

```python
# recursão sobre achar o maior valor de uma lista 
def maior_valor(lista):
    if  len(lista)== 1:
        return lista[0]
    elif lista[0] > maior_valor(lista[1:]) :
        return lista[0]
    else: 
        return maior_valor(lista[1:])

maior_valor([15,11,16])
```

# Codigo de exemplo quicksort

```python
# tornando um array ordenado
def quicksort(array):
    if len(array) < 2:
        return array
    else:
        pivo = array[0]
        menores = [i for i in array[1:] if i <= pivo]
        maiores = [i for i in array[1:] if i > pivo]
        return quicksort(menores) + [pivo] + quicksort(maiores)
quicksort([3,10,11,4,7])
        
        
```

# Ordenação de numeros em ordem decrescente 

```python
def quicksort(array):
    if len(array) < 2 :
        return array 
    else:
        pivo= array[0]
        menores = [i for i in array[1:] if i > pivo]
        maiores = [i for i in array[1:] if i <= pivo]
        return quicksort(menores) + [pivo] + quicksort(maiores)
quicksort([3,10,11,4,7])
```
